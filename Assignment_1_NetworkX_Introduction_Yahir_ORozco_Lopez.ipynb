# Part 2 - Create a Simple Graph

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
G.add_edge(1, 2)
G.add_edge(2, 3)
G.add_edge(3, 1)
nx.draw(G, with_labels=True)
plt.show()
```

1. The numbers 1, 2, and 3 represent the nodes
2. The lines connecting the nodes represent the edges
3. This graph is undirected
4. I can tell due to how none of the edges have any arrows to indicate direction

# Part 3 - Explore the Graph

```python
print("Nodes: ", G.nodes())
print("Edges: ", G.edges())
print("Degree of each node: ", dict(G.degree()))

# getting number of nodes and edges
print("# of nodes: ", len(list(G.nodes())))
print("# of edges: ", len(list(G.edges())))
```

1. There are 3 nodes in the graph
2. There are 3 edges in the graph
3. The degree of each node is 2
4. All nodes have the same degree because each node connects to the same amount of nodes (2).
5. The **degree of a node** represents the amount of nodes that a specific node connects to.

# Part 4 - Compare Graph Types

```python
DG = nx.DiGraph()

DG.add_edge(1, 2)
DG.add_edge(2, 3)
DG.add_edge(3, 1)

nx.draw(DG, with_labels=True, arrows=True)
plt.show()
```

1. The difference between *nx.Graph()* and *nx.DiGraph()* is that the first method call creates an undirected graph while the second creates a direction graph.
2. I notice that the *.DiGraph()* method creates arrows pertaining to what is passed in the arguments for the method.
3. The direction of an edge represents the relationship between nodes.
4. A real-world network where direction would matter could include a network of train tracks exist between cities and states.

# Part 5 - Create Your Own Network
- Network that represents family tree


```python
fam_DG = nx.DiGraph()

# immediate family
fam_DG.add_node("mom")
fam_DG.add_node("dad")
fam_DG.add_node("yahir")
fam_DG.add_node("osvaldo")
fam_DG.add_node("pablo")
fam_DG.add_node("briana")

# parents and siblings
fam_DG.add_edge("mom", "yahir")
fam_DG.add_edge("mom", "osvaldo")
fam_DG.add_edge("mom", "pablo")
fam_DG.add_edge("mom", "briana")

fam_DG.add_edge("dad", "yahir")
fam_DG.add_edge("dad", "osvaldo")
fam_DG.add_edge("dad", "pablo")
fam_DG.add_edge("dad", "briana")

# osvaldo's family
fam_DG.add_edge("osvaldo", "luna")
fam_DG.add_edge("osvaldo", "isaiah")
fam_DG.add_edge("osvaldo", "julian")

# pablo's family
fam_DG.add_edge("pablo", "andrea")
fam_DG.add_edge("pablo", "alezar")
fam_DG.add_edge("pablo", "viviana")
fam_DG.add_edge("pablo", "lalo")

# briana's family
fam_DG.add_edge("briana", "aurora")
fam_DG.add_edge("briana", "aaron")
fam_DG.add_edge("briana", "alanna")
fam_DG.add_edge("briana", "amara")

"""
below code was taken from
https://networkx.org/documentation/latest/auto_examples/basic/plot_basic_directed.html
"""

# group nodes by column
left_nodes = ["mom", "dad"]
middle_nodes = ["yahir", "osvaldo", "pablo", "briana"]
right_nodes = ["luna", "isaiah", "julian", "andrea", "alezar", "viviana",
              "lalo", "aurora", "aaron", "alanna", "amara"]

# set the position according to column (x-coord)
pos = {n: (0, i) for i, n in enumerate(left_nodes)}
pos.update({n: (1, i + 0.5) for i, n in enumerate(middle_nodes)})
pos.update({n: (2, i + 0.5) for i, n in enumerate(right_nodes)})


nx.draw(fam_DG, pos, with_labels=True, arrows=True)
plt.show()
```
